# Notebook 01: Data Forensics Audit & Pipeline Hygiene
**Objective**: Investigate raw data inconsistencies, deduplicate failed/pending payments, resolve synthetic agent identities, and standardize timezones.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Load Raw Payments Data
df_pay_raw = pd.read_csv('../data/raw/payments.csv')
print('Raw Payments Total Count:', len(df_pay_raw))
print('Raw Payments Gross Amount: INR', df_pay_raw['amount'].sum())
display(df_pay_raw['payment_status'].value_counts())
```

```python
# 2. Filter Non-SUCCESS Payments and Deduplicate References
df_pay_success = df_pay_raw[df_pay_raw['payment_status'] == 'SUCCESS'].copy()
df_pay_clean = df_pay_success.drop_duplicates(subset=['payment_reference'], keep='first').copy()

print('Clean Net Verified Payments Count:', len(df_pay_clean))
print('Clean Net Verified Recovery Capital: INR', df_pay_clean['amount'].sum())
print('Gross Over-Reporting Bias: INR', df_pay_raw['amount'].sum() - df_pay_clean['amount'].sum())
```
