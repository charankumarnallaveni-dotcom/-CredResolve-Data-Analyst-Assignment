# Notebook 02: Independent 11% Claim Validation
**Objective**: Evaluate the business claim 'Recovery improved by 11% MoM' across 7 complete months (Jan–Jul 2026) and partial August 2026.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df_mp = pd.read_csv('../data/golden/monthly_performance.csv')
df_bridge = pd.read_csv('../reports/recovery_truth_bridge.csv')
display(df_mp[['month', 'accounts', 'recovery_amount_inr', 'recovery_rate_pct', 'recovery_per_account_inr']])
```

```python
# Calculate MoM Recovery Rate Changes
df_mp['recovery_rate_mom_change'] = df_mp['recovery_rate_pct'].diff()
df_mp['recovery_per_acc_mom_pct'] = df_mp['recovery_per_account_inr'].pct_change() * 100
display(df_mp[['month', 'recovery_rate_pct', 'recovery_rate_mom_change', 'recovery_per_account_inr', 'recovery_per_acc_mom_pct']])
```

```python
# Plot Clean Recovery Rate Trend across Complete Months
df_complete = df_mp[df_mp['month'] <= '2026-07']
plt.figure(figsize=(10, 4))
plt.plot(df_complete['month'], df_complete['recovery_rate_pct'], marker='o', color='crimson', linewidth=2.5)
plt.title('Clean Independent Recovery Rate (%) Across 7 Complete Months (Jan - Jul 2026)')
plt.ylabel('Recovery Rate (%)')
plt.grid(True, linestyle=':')
plt.show()
```
