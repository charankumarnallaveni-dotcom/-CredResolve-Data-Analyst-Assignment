```python
# Portfolio DPD Shift Analysis (Pre vs Post April Cliff Drop)
df_pay = pd.read_csv('../data/golden/fct_payments.csv')
df_acc = pd.read_csv('../data/golden/dim_accounts.csv')
df_merged = df_pay.merge(df_acc, on='account_id')
df_merged['period'] = np.where(df_merged['event_at_ist'] < '2026-04-01', 'Pre-April (Jan-Mar)', 'Post-April (Apr-Jul)')
dpd_shift = pd.crosstab(df_merged['period'], df_merged['dpd'] > 60, normalize='index') * 100
print('Percentage of Payments from Accounts > 60 DPD:')
display(dpd_shift)
```
