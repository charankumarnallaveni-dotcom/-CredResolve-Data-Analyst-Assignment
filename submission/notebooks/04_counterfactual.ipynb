# Notebook 04: Targeting Strategy Counterfactual Analysis (DiD Model)
**Objective**: Estimate counterfactual recovery if strategy version 'v2' had been maintained in April 2026.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df_cf = pd.read_csv('../data/golden/counterfactual_results.csv')
display(df_cf)
```

```python
# Calculate DiD Estimator Summary
pre_treat = 20209.67
post_treat = 17447.87
pre_ctrl = 41570.98
post_ctrl = 35316.38
did_est = (post_treat - pre_treat) - (post_ctrl - pre_ctrl)
print(f'DiD Estimator: INR {did_est:.2f} per account')
print('Interpretation: Strategy v3 mitigated collection yield decay by INR 3,492.80 per account compared to control.')
```
