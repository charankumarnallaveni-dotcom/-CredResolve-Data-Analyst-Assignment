# Notebook 05: ₹10 Crore Investment Financial Model
**Objective**: Evaluate 6 candidate investment options under Downside, Base, and Upside scenarios.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df_inv = pd.read_csv('../data/golden/investment_comparison.csv')
display(df_inv[['option_name', 'base_incremental_recovery_12m', 'base_roi_pct', 'break_even_month', 'confidence_level', 'recommendation_status']])
```

```python
# Plot Base ROI Comparison
plt.figure(figsize=(10, 4))
plt.barh(df_inv['option_name'], df_inv['base_roi_pct'], color=['#059669' if r > 60 else '#94A3B8' for r in df_inv['base_roi_pct']])
plt.title('12-Month Base Net ROI (%) Across 6 Investment Options')
plt.xlabel('Base ROI (%)')
plt.grid(axis='x', linestyle=':')
plt.show()
```
